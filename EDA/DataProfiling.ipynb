```python
# Cell gọi dữ liệu
import pandas as pd 
import os
from datetime import datetime
import os
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

print('Đang khai báo dữ liệu')

# Dữ liệu thô ban đầu
file_path = os.path.join("..", "Data", "train.csv")

df = pd.read_csv(file_path)

print('Khai báo thành công')
```

```python
## 1. Tổng quan Dataset
```

```python
print("=" * 50)
print("KÍCH THƯỚC DATASET")
print("=" * 50)
print(f"Số dòng (bài hát): {df.shape[0]}")
print(f"Số cột (đặc trưng): {df.shape[1]}")
print(f"\nTên các cột:")
for col in df.columns:
    print(f"  - {col}")
```

```python
## 2. Xem mẫu dữ liệu
```

```python
print("5 dòng đầu:")
display(df.head())

print("\n5 dòng cuối:")
display(df.tail())
```

```python
# Cấu trúc lưu ảnh hoặc file csv hoặc dữ liệu nào cần thiết
# Đây là cấu trúc lưu file csv, nhưng hầu như công đoạn này lưu ảnh

now = datetime.now().strftime("%Y%m%d_%H%M%S")

filename = f"{now}_data_profiling.csv"
filepath = os.path.join("..", "Notebooks", "EDA", "DataProfiling", filename)
os.makedirs(os.path.dirname(filepath), exist_ok=True)

df.to_csv(filepath, index=False)

print("Saved:", filepath)
```

```python
## 3. Kiểu dữ liệu từng cột
```

```python
print("Kiểu dữ liệu từng cột:")
print(df.dtypes)
print(f"\nTổng quan:")
df.info()
```

```python
## 4. Thống kê mô tả
```

```python
print("Thống kê mô tả các cột số:")
display(df.describe())
```

```python
## 5. Kiểm tra Missing Values
```

```python
missing = df.isnull().sum()
missing_pct = (df.isnull().sum() / len(df) * 100).round(2)

missing_df = pd.DataFrame({
    'Số lượng thiếu': missing,
    'Phần trăm (%)': missing_pct
}).sort_values('Số lượng thiếu', ascending=False)

print("Thống kê giá trị thiếu:")
display(missing_df[missing_df['Số lượng thiếu'] > 0])

print(f"\nTổng số ô thiếu dữ liệu: {df.isnull().sum().sum()}")
```

```python
## 6. Kiểm tra Duplicate
```

```python
dup_count = df.duplicated().sum()
print(f"Số dòng bị trùng lặp: {dup_count}")

if dup_count > 0:
    print("\nCác dòng bị trùng:")
    display(df[df.duplicated(keep=False)])
else:
    print("Không có dòng nào bị trùng lặp.")
```

```python
## 7. Phân phối nhãn (Class)
```

```python
print("Phân phối các thể loại nhạc (Class):")
class_counts = df['Class'].value_counts().sort_index()
display(class_counts.to_frame('Số lượng'))

print(f"\nTổng số thể loại: {df['Class'].nunique()}")

# Vẽ biểu đồ
plt.figure(figsize=(10, 5))
sns.countplot(data=df, x='Class', palette='Set2', order=sorted(df['Class'].unique()))
plt.title('Phân phối thể loại nhạc trong tập train')
plt.xlabel('Class (Thể loại)')
plt.ylabel('Số lượng bài hát')
plt.tight_layout()
plt.savefig('../EDA/class_distribution.png', dpi=150)
plt.show()
print("Đã lưu biểu đồ: class_distribution.png")
```

```python
## 8. Phân phối các đặc trưng số
```

```python
numeric_cols = ['Popularity', 'danceability', 'energy', 'loudness',
                'speechiness', 'acousticness', 'instrumentalness',
                'liveness', 'valence', 'tempo']

df[numeric_cols].hist(bins=30, figsize=(15, 10), color='steelblue', edgecolor='white')
plt.suptitle('Phân phối các đặc trưng số', fontsize=14)
plt.tight_layout()
plt.savefig('../EDA/feature_distributions.png', dpi=150)
plt.show()
print("Đã lưu biểu đồ: feature_distributions.png")
```
