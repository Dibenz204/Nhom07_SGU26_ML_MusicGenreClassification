```python
# Cell: Khai báo và chuẩn bị dữ liệu cho train/val/test

import pandas as pd
import os
import numpy as np

# 1. Load dữ liệu
print("Loading data...")

train_path = os.path.join("Data", "DataCleaned", "data_train_cleaned.csv")
val_path = os.path.join("Data", "DataCleaned", "validation_cleaned.csv")
test_path = os.path.join("Data", "DataCleaned", "test_cleaned.csv")

df_train = pd.read_csv(train_path)
df_val = pd.read_csv(val_path)
df_test = pd.read_csv(test_path)

print(f"Train shape: {df_train.shape}")
print(f"Val shape: {df_val.shape}")
print(f"Test shape: {df_test.shape}")

# 2. Chuẩn bị X, y
drop_cols = ['Id', 'Artist Name', 'Track Name']

# X_train, y_train
existing_drop_train = [col for col in drop_cols if col in df_train.columns]
X_train = df_train.drop(columns=['Class'] + existing_drop_train)
y_train = df_train['Class']

# X_val, y_val
existing_drop_val = [col for col in drop_cols if col in df_val.columns]
X_val = df_val.drop(columns=['Class'] + existing_drop_val)
y_val = df_val['Class']

# X_test (không có nhãn)
existing_drop_test = [col for col in drop_cols if col in df_test.columns]
X_test = df_test.drop(columns=existing_drop_test)

# 3. Bỏ feature 'key' vì đã có key_sin và key_cos
X_train = X_train.drop(columns=['key'])
X_val = X_val.drop(columns=['key'])
X_test = X_test.drop(columns=['key'])

print(f"\n✅ After dropping 'key':")
print(f"X_train shape: {X_train.shape}")
print(f"X_val shape: {X_val.shape}")
print(f"X_test shape: {X_test.shape}")

# 4. Lưu feature names để dùng sau
feature_names = X_train.columns.tolist()
print(f"\n📋 Feature names ({len(feature_names)} features):")
print(feature_names)
```

```python
# Cell: So sánh features của 3 tập

print("="*55)
print("COMPARE FEATURES ACROSS TRAIN/VAL/TEST")
print("="*55)

print("\n=== TRAIN columns ===")
print(f"Number of features: {len(X_train.columns)}")
print(X_train.columns.tolist())

print("\n=== VAL columns ===")
print(f"Number of features: {len(X_val.columns)}")
print(X_val.columns.tolist())

print("\n=== TEST columns ===")
print(f"Number of features: {len(X_test.columns)}")
print(X_test.columns.tolist())

# Kiểm tra features có khớp nhau không
print("\n=== FEATURE ALIGNMENT CHECK ===")
train_features = set(X_train.columns)
val_features = set(X_val.columns)
test_features = set(X_test.columns)

print(f"✅ Features in TRAIN only: {train_features - val_features - test_features}")
print(f"✅ Features in VAL only: {val_features - train_features - test_features}")
print(f"✅ Features in TEST only: {test_features - train_features - val_features}")

if train_features == val_features == test_features:
    print("\n🎉 All datasets have the same features!")
else:
    print("\n⚠️ Warning: Feature mismatch detected!")

# So sánh mean giữa train và test
print("\n=== TRAIN mean vs TEST mean (Top 10 differences) ===")
compare_test = pd.DataFrame({
    'train_mean': X_train.mean(),
    'test_mean': X_test.mean(),
    'diff_pct': ((X_train.mean() - X_test.mean()) / X_train.mean() * 100).round(2)
})
compare_test['abs_diff'] = compare_test['diff_pct'].abs()
compare_test = compare_test.sort_values('abs_diff', ascending=False)
print(compare_test.head(10))

# So sánh mean giữa train và val
print("\n=== TRAIN mean vs VAL mean (Top 10 differences) ===")
compare_val = pd.DataFrame({
    'train_mean': X_train.mean(),
    'val_mean': X_val.mean(),
    'diff_pct': ((X_train.mean() - X_val.mean()) / X_train.mean() * 100).round(2)
})
compare_val['abs_diff'] = compare_val['diff_pct'].abs()
compare_val = compare_val.sort_values('abs_diff', ascending=False)
print(compare_val.head(10))

# Kiểm tra nhanh xem đã bỏ 'key' thành công chưa
print("\n=== VERIFY 'key' REMOVAL ===")
print(f"'key' in TRAIN: {'key' in X_train.columns}")
print(f"'key' in VAL: {'key' in X_val.columns}")
print(f"'key' in TEST: {'key' in X_test.columns}")
```
